# MIRFLICKR 25K Transfer-Tuning Results

This notebook summarizes latest tuning outputs (trial-level, epoch-level, and run history) under `/app/outputs/training`.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

base = Path('/app/outputs/training')
runs = sorted([p for p in base.iterdir() if p.is_dir()])
latest = runs[-1]
print('Latest run:', latest)
summary = json.loads((latest / 'summary.json').read_text())
trials = pd.read_csv(latest / 'trial_results.csv')
epoch_path = latest / 'trial_epoch_logs.csv'
epochs = pd.read_csv(epoch_path) if epoch_path.exists() else pd.DataFrame()
summary
```

```python
trials.head(10)
```

```python
x_col = 'n_estimators' if 'n_estimators' in trials.columns else 'trial_id'
ax = trials.sort_values('f1_weighted', ascending=False).head(10).plot(
    x=x_col, y='f1_weighted', kind='bar', figsize=(10,4), legend=False, color='steelblue'
)
ax.set_title('Top 10 Trials by Weighted F1')
ax.set_ylabel('f1_weighted')
plt.tight_layout()
```

```python
if not epochs.empty:
    best_trial = int(trials.sort_values('f1_weighted', ascending=False).iloc[0]['trial_id'])
    ed = epochs[epochs['trial_id'] == best_trial].sort_values('epoch')
    ax = ed.plot(x='epoch', y=['accuracy','f1_weighted'], figsize=(8,4), marker='o')
    ax.set_title(f'Best Trial Learning Curve (trial_id={best_trial})')
    plt.tight_layout()
else:
    print('No epoch logs found for latest run.')
```
